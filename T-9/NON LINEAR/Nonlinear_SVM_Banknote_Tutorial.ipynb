# Nonlinear Support Vector Machine Tutorial: EEG Eye State Detection

This notebook uses the real-life **UCI EEG Eye State** dataset. A 14-channel EEG headset recorded brain activity while the person's eyes were open or closed. The task is to predict the eye state from the EEG signal.

- Official dataset page: https://archive.ics.uci.edu/dataset/264/eeg+eye+state
- Direct download: https://archive.ics.uci.edu/static/public/264/eeg+eye+state.zip
- Original observations: 14,980 (a 5,000-row stratified subset is used so tuning runs quickly)
- Predictors: 14 continuous EEG channels (`AF3, F7, F3, FC5, T7, P7, O1, O2, P8, T8, FC6, F4, F8, AF4`)
- Target: `class` (`0` = eyes open, `1` = eyes closed)
- Local fallback: `EEG_Eye_State_NonLinear_SVM_Sample.xlsx` (500-row sample) if the download is unavailable

This tutorial focuses exclusively on nonlinear SVMs using polynomial, radial basis function (RBF), and sigmoid kernels. It includes preprocessing, kernel comparison, kernel-specific hyperparameter tuning, evaluation, and visualization.

```python
# 1. IMPORT LIBRARIES
import os
from io import BytesIO, StringIO
from urllib.request import urlopen
from zipfile import ZipFile

from scipy.io import arff

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold
from sklearn.decomposition import PCA
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report,
    ConfusionMatrixDisplay, RocCurveDisplay, PrecisionRecallDisplay
)

sns.set_theme(style='whitegrid', palette='Set2')
RANDOM_STATE = 42
MAX_ROWS = 5000  # keeps SVM grid search fast; set to None to use all rows
```

```python
# 2. LOAD THE REAL-LIFE DATASET (UCI download, with local-file fallback)
DATA_URL = 'https://archive.ics.uci.edu/static/public/264/eeg+eye+state.zip'
EEG_COLUMNS = ['AF3', 'F7', 'F3', 'FC5', 'T7', 'P7', 'O1', 'O2',
               'P8', 'T8', 'FC6', 'F4', 'F8', 'AF4']
LOCAL_CANDIDATES = [
    'EEG_Eye_State_NonLinear_SVM_Sample.xlsx',
    'EEG_Eye_State_NonLinear_SVM_Sample (1).xlsx', 'non_linear.xlsx'
]

try:
    with urlopen(DATA_URL, timeout=60) as response:
        zip_bytes = BytesIO(response.read())

    with ZipFile(zip_bytes) as archive:
        arff_name = next(n for n in archive.namelist() if n.lower().endswith('.arff'))
        arff_text = archive.read(arff_name).decode('utf-8')

    data, meta = arff.loadarff(StringIO(arff_text))
    df = pd.DataFrame(data)
    df.columns = EEG_COLUMNS + ['class']
    df['class'] = df['class'].apply(
        lambda v: v.decode('utf-8') if isinstance(v, bytes) else v
    ).astype(int)
    print('Loaded full EEG Eye State dataset from UCI.')

except Exception as error:
    print('UCI download unavailable:', error)
    local_file = next((f for f in LOCAL_CANDIDATES if os.path.exists(f)), None)
    if local_file is None:
        raise FileNotFoundError(
            'Download failed and no local sample found. Place the EEG sample xlsx '
            f'next to the notebook. Accepted names: {LOCAL_CANDIDATES}'
        )
    df = pd.read_excel(local_file)
    df = df.rename(columns={df.columns[-1]: 'class'})  # last column = eye state label
    print(f'Loaded local sample file: {local_file}')

print('Dataset shape:', df.shape)
display(df.head())
```

```python
# 3. INSPECT DATA QUALITY
print('Data types:\n', df.dtypes)
print('\nMissing values by column:\n', df.isna().sum())
print('\nDuplicate rows:', int(df.duplicated().sum()))
print('\nClass counts:\n', df['class'].value_counts().sort_index())
print('\nClass proportions:\n', df['class'].value_counts(normalize=True).sort_index().round(4))
display(df.describe().T)
```

```python
# 4. PREPROCESS THE DATA
# Remove exact duplicate observations before splitting the dataset.
df = df.drop_duplicates().reset_index(drop=True)

feature_columns = [col for col in df.columns if col != 'class']

# Force predictors and target to numeric data. Invalid values become NaN.
df[feature_columns] = df[feature_columns].apply(pd.to_numeric, errors='coerce')
df['class'] = pd.to_numeric(df['class'], errors='coerce')

# Median imputation is included for robustness even though no missing values are expected.
df[feature_columns] = df[feature_columns].fillna(df[feature_columns].median())
df = df.dropna(subset=['class']).copy()
df['class'] = df['class'].astype(int)

# EEG recordings contain a few extreme sensor artefacts (spikes far outside the normal range).
# Remove rows lying beyond 3 x IQR from the quartiles of any channel.
q1 = df[feature_columns].quantile(0.25)
q3 = df[feature_columns].quantile(0.75)
iqr = q3 - q1
inside_range = ((df[feature_columns] >= q1 - 3 * iqr) &
                (df[feature_columns] <= q3 + 3 * iqr)).all(axis=1)
print('Rows removed as sensor artefacts:', int((~inside_range).sum()))
df = df[inside_range].reset_index(drop=True)

# Stratified subsample so the kernel grid searches finish quickly.
if MAX_ROWS is not None and len(df) > MAX_ROWS:
    df, _ = train_test_split(df, train_size=MAX_ROWS, stratify=df['class'],
                             random_state=RANDOM_STATE)
    df = df.reset_index(drop=True)

X = df[feature_columns]
y = df['class']

print('Shape after preprocessing:', df.shape)
print('Remaining missing values:', int(df.isna().sum().sum()))
print('Target labels:', sorted(y.unique()))
```

```python
# 8. CREATE STRATIFIED TRAINING AND TEST SETS
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=RANDOM_STATE, stratify=y
)

print('Training set:', X_train.shape)
print('Testing set:', X_test.shape)
print('\nTraining class proportions:\n', y_train.value_counts(normalize=True).sort_index().round(4))
print('\nTesting class proportions:\n', y_test.value_counts(normalize=True).sort_index().round(4))
```

## Nonlinear SVM kernels

Each kernel name is passed as a string to `SVC(kernel=kernel_name)`.

- **Polynomial (`'poly'`)** models curved relationships through polynomial feature combinations.
- **RBF (`'rbf'`)** creates flexible local decision regions based on radial similarity.
- **Sigmoid (`'sigmoid'`)** uses a hyperbolic tangent transformation similar to a neural activation function.

```python
# 9. IMPLEMENT BASELINE NONLINEAR SVM KERNELS
kernel_names = ['poly', 'rbf', 'sigmoid']
baseline_models = {}
baseline_rows = []

for kernel_name in kernel_names:
    # The kernel name is passed as a string to SVC.
    model = Pipeline([
        ('scaler', StandardScaler()),
        ('svc', SVC(kernel=kernel_name, C=1.0, gamma='scale'))
    ])
    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)
    y_score = model.decision_function(X_test)
    baseline_models[kernel_name] = model

    baseline_rows.append({
        'Kernel': kernel_name,
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred, zero_division=0),
        'Recall': recall_score(y_test, y_pred, zero_division=0),
        'F1-score': f1_score(y_test, y_pred, zero_division=0),
        'ROC-AUC': roc_auc_score(y_test, y_score),
        'Support Vectors': int(model.named_steps['svc'].n_support_.sum())
    })

baseline_results = pd.DataFrame(baseline_rows).sort_values(
    'F1-score', ascending=False
).reset_index(drop=True)
display(baseline_results.round(4))
```

```python
# 10. VISUALIZE BASELINE KERNEL COMPARISON
baseline_long = baseline_results.melt(
    id_vars='Kernel',
    value_vars=['Accuracy', 'Precision', 'Recall', 'F1-score', 'ROC-AUC'],
    var_name='Metric', value_name='Score'
)

plt.figure(figsize=(10, 6))
sns.barplot(data=baseline_long, x='Kernel', y='Score', hue='Metric')
plt.ylim(0, 1.05)
plt.title('Baseline Nonlinear SVM Kernel Comparison')
plt.ylabel('Test Score')
plt.tight_layout()
plt.show()
```

## Kernel-specific hyperparameter tuning

Separate searches are used because each nonlinear kernel has different relevant parameters:

- `C`: penalty for classification errors
- `gamma`: influence radius of individual training points
- `degree`: polynomial degree, used only by the polynomial kernel
- `coef0`: independent kernel term used by polynomial and sigmoid kernels

Scaling stays inside the pipeline so that every cross-validation fold learns preprocessing only from its own training portion.

```python
# 11. DEFINE KERNEL-SPECIFIC HYPERPARAMETER GRIDS
kernel_parameter_grids = {
    'poly': {
        'svc__C': [0.1, 1, 10],
        'svc__gamma': ['scale', 0.1, 1],
        'svc__degree': [2, 3, 4],
        'svc__coef0': [0.0, 1.0]
    },
    'rbf': {
        'svc__C': [0.1, 1, 10, 100],
        'svc__gamma': ['scale', 0.01, 0.1, 1]
    },
    'sigmoid': {
        'svc__C': [0.1, 1, 10],
        'svc__gamma': ['scale', 0.01, 0.1],
        'svc__coef0': [-1.0, 0.0, 1.0]
    }
}

for kernel_name, parameter_grid in kernel_parameter_grids.items():
    combinations = int(np.prod([len(values) for values in parameter_grid.values()]))
    print(f'{kernel_name:<7}: {combinations} hyperparameter combinations')
```

```python
# 12. TUNE EACH NONLINEAR KERNEL WITH FIVE-FOLD GRID SEARCH
cv_strategy = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
searches = {}
tuned_models = {}
tuning_summary_rows = []
all_cv_results = []

for kernel_name in kernel_names:
    pipeline = Pipeline([
        ('scaler', StandardScaler()),
        ('svc', SVC(kernel=kernel_name))
    ])

    search = GridSearchCV(
        estimator=pipeline,
        param_grid=kernel_parameter_grids[kernel_name],
        scoring='f1',
        cv=cv_strategy,
        n_jobs=-1,
        refit=True,
        return_train_score=True
    )
    search.fit(X_train, y_train)

    searches[kernel_name] = search
    tuned_models[kernel_name] = search.best_estimator_
    tuning_summary_rows.append({
        'Kernel': kernel_name,
        'Best CV F1-score': search.best_score_,
        'Best Parameters': search.best_params_
    })

    kernel_cv = pd.DataFrame(search.cv_results_)
    kernel_cv['Kernel'] = kernel_name
    all_cv_results.append(kernel_cv)

    print(f'\n{kernel_name.upper()} kernel')
    print('Best parameters:', search.best_params_)
    print('Best mean CV F1-score:', round(search.best_score_, 4))

tuning_summary = pd.DataFrame(tuning_summary_rows).sort_values(
    'Best CV F1-score', ascending=False
).reset_index(drop=True)
display(tuning_summary)
```

```python
# 13. VISUALIZE CROSS-VALIDATION PERFORMANCE
combined_cv_results = pd.concat(all_cv_results, ignore_index=True)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(
    data=combined_cv_results, x='Kernel', y='mean_test_score',
    hue='Kernel', legend=False, ax=axes[0]
)
axes[0].set_title('Distribution of Grid-Search CV F1-scores')
axes[0].set_xlabel('Kernel')
axes[0].set_ylabel('Mean Five-Fold CV F1-score')

sns.barplot(
    data=tuning_summary, x='Kernel', y='Best CV F1-score',
    hue='Kernel', legend=False, ax=axes[1]
)
axes[1].set_ylim(0, 1.02)
axes[1].set_title('Best Cross-Validation Score per Kernel')
axes[1].set_ylabel('Best Mean CV F1-score')

plt.tight_layout()
plt.show()
```

```python
# 14. EVALUATE EACH TUNED KERNEL ON THE UNSEEN TEST SET
tuned_rows = []
kernel_predictions = {}
kernel_scores = {}

for kernel_name, model in tuned_models.items():
    y_pred = model.predict(X_test)
    y_score = model.decision_function(X_test)
    kernel_predictions[kernel_name] = y_pred
    kernel_scores[kernel_name] = y_score

    tuned_rows.append({
        'Kernel': kernel_name,
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred, zero_division=0),
        'Recall': recall_score(y_test, y_pred, zero_division=0),
        'F1-score': f1_score(y_test, y_pred, zero_division=0),
        'ROC-AUC': roc_auc_score(y_test, y_score),
        'Support Vectors': int(model.named_steps['svc'].n_support_.sum())
    })

tuned_results = pd.DataFrame(tuned_rows).sort_values(
    'F1-score', ascending=False
).reset_index(drop=True)
display(tuned_results.round(4))
```

```python
# 15. COMPARE BASELINE AND TUNED KERNELS
comparison = baseline_results[['Kernel', 'F1-score']].merge(
    tuned_results[['Kernel', 'F1-score']],
    on='Kernel', suffixes=(' Baseline', ' Tuned')
)
comparison_long = comparison.melt(
    id_vars='Kernel', var_name='Model', value_name='F1-score'
)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(
    data=comparison_long, x='Kernel', y='F1-score', hue='Model', ax=axes[0]
)
axes[0].set_ylim(0, 1.05)
axes[0].set_title('Baseline vs Tuned Test F1-score')

tuned_long = tuned_results.melt(
    id_vars='Kernel',
    value_vars=['Accuracy', 'Precision', 'Recall', 'F1-score', 'ROC-AUC'],
    var_name='Metric', value_name='Score'
)
sns.barplot(data=tuned_long, x='Kernel', y='Score', hue='Metric', ax=axes[1])
axes[1].set_ylim(0, 1.05)
axes[1].set_title('Tuned Kernel Test Metrics')
axes[1].legend(loc='lower right')

plt.tight_layout()
plt.show()
```

```python
# 16. VISUALIZE CONFUSION MATRICES FOR ALL TUNED KERNELS
fig, axes = plt.subplots(1, 3, figsize=(16, 4.5))

for kernel_name, ax in zip(kernel_names, axes):
    ConfusionMatrixDisplay.from_predictions(
        y_test, kernel_predictions[kernel_name],
        display_labels=['Eyes open', 'Eyes closed'],
        cmap='Blues', colorbar=False, ax=ax
    )
    ax.set_title(f'{kernel_name.upper()} Confusion Matrix')

plt.tight_layout()
plt.show()
```

```python
# 17. COMPARE ROC AND PRECISION-RECALL CURVES
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

for kernel_name in kernel_names:
    RocCurveDisplay.from_predictions(
        y_test, kernel_scores[kernel_name],
        name=kernel_name.upper(), ax=axes[0]
    )
    PrecisionRecallDisplay.from_predictions(
        y_test, kernel_scores[kernel_name],
        name=kernel_name.upper(), ax=axes[1]
    )

axes[0].plot([0, 1], [0, 1], 'k--', label='Random classifier')
axes[0].set_title('ROC Curve Comparison')
axes[0].legend()
axes[1].set_title('Precision-Recall Curve Comparison')

plt.tight_layout()
plt.show()
```

```python
# 18. PRINT DETAILED CLASSIFICATION REPORTS
for kernel_name in kernel_names:
    print(f'\n--- {kernel_name.upper()} KERNEL ---')
    print('Best parameters:', searches[kernel_name].best_params_)
    print(classification_report(
        y_test, kernel_predictions[kernel_name],
        target_names=['Eyes open', 'Eyes closed'], digits=4
    ))
```

```python
# 19. VISUALIZE NONLINEAR DECISION BOUNDARIES
# The data has 14 features, so a 2-component PCA projection is used for visualization only.
# Evaluated models above use all 14 EEG channels.
vis_prep = Pipeline([
    ('scaler', StandardScaler()),
    ('pca', PCA(n_components=2, random_state=RANDOM_STATE))
])
X_vis_train = vis_prep.fit_transform(X_train)
X_vis_all = vis_prep.transform(X)

x_min, x_max = np.percentile(X_vis_all[:, 0], [1, 99])
y_min, y_max = np.percentile(X_vis_all[:, 1], [1, 99])
xx, yy = np.meshgrid(
    np.linspace(x_min, x_max, 300),
    np.linspace(y_min, y_max, 300)
)
grid_points = np.c_[xx.ravel(), yy.ravel()]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for kernel_name, ax in zip(kernel_names, axes):
    best_svc_parameters = {
        key.replace('svc__', ''): value
        for key, value in searches[kernel_name].best_params_.items()
    }
    visual_model = SVC(kernel=kernel_name, **best_svc_parameters)
    visual_model.fit(X_vis_train, y_train)
    decision_values = visual_model.decision_function(grid_points).reshape(xx.shape)

    ax.contourf(xx, yy, decision_values > 0, alpha=0.22, cmap='coolwarm')
    ax.contour(
        xx, yy, decision_values, levels=[-1, 0, 1],
        colors=['gray', 'black', 'gray'], linestyles=['--', '-', '--']
    )
    ax.scatter(
        X_vis_all[:, 0], X_vis_all[:, 1],
        c=y, cmap='coolwarm', edgecolor='none', s=12, alpha=0.5
    )
    ax.set_xlim(x_min, x_max)
    ax.set_ylim(y_min, y_max)
    ax.set_title(f'{kernel_name.upper()} Decision Boundary (PCA 2D)')
    ax.set_xlabel('Principal Component 1')
    ax.set_ylabel('Principal Component 2')

plt.tight_layout()
plt.show()
```

```python
# 20. FINAL SUMMARY
best_kernel = tuned_results.iloc[0]['Kernel']
best_row = tuned_results.iloc[0]

print('Nonlinear SVM tutorial completed.')
print('Compared kernels:', ', '.join(kernel_names))
print('Best test-set kernel:', best_kernel)
print('Best parameters:', searches[best_kernel].best_params_)
print('Test accuracy:', round(best_row['Accuracy'], 4))
print('Test F1-score:', round(best_row['F1-score'], 4))
print('Test ROC-AUC:', round(best_row['ROC-AUC'], 4))
print('The test set was not used during hyperparameter selection.')
```
